```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('color1.jpg', 0)
template = cv2.imread('color1.jpg', 0)
w, h = template.shape[::-1]
o = img.copy()

# Apply template Matching
res = cv2.matchTemplate(img, template, cv2.TM_CCOEFF_NORMED)
threshold = 0.8
loc = np.where(res >= threshold)

# Draw rectangles around matches
for pt in zip(*loc[::-1]):
    cv2.rectangle(img, pt, (pt[0] + w, pt[1] + h), (255, 255, 255), 2)
    
output_rgb = cv2.cvtColor(img, cv2.COLOR_GRAY2RGB)
print(output_rgb.shape)
    
cv2.imshow('Faces Detected', output_rgb)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.subplot(1, 3, 1)
plt.imshow(o, cmap="gray")
plt.title("Original")
plt.axis("off")

# الصورة الرمادية
plt.subplot(1, 3, 2)
plt.imshow(template, cmap="gray")
plt.title("Template")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(img, cmap="gray")
plt.title("Result")
plt.axis("off")

plt.show()
```
