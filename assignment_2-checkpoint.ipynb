```python
import cv2
import matplotlib.pyplot as plt
import os
```

```python
IMAGE_PATH = "scene.jpg"
```

```python
img_bgr = cv2.imread(IMAGE_PATH)

print("Image shape (H, W, Channels):", img_bgr.shape)

if img_bgr is None:

    raise FileNotFoundError(f"Could not load image at {IMAGE_PATH}")

```

```python
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))

plt.imshow(img_rgb)

plt.title("Original Scene (RGB)")

plt.axis("off")

plt.show()

```

```python
x_start, y_start = 730, 480

x_end, y_end = 1180, 830

roi = img_rgb[y_start:y_end, x_start:x_end]

plt.figure(figsize=(5, 5))

plt.imshow(roi)

plt.title("Cropped Object (ROI) - Bananas")

plt.axis("off")

plt.show()
```

```python
roi_resized = cv2.resize(roi, (256, 256), interpolation=cv2.INTER_LINEAR)

plt.figure(figsize=(4, 4))

plt.imshow(roi_resized)

plt.title("Resized ROI (256x256)")

plt.axis("off")

plt.show()

print("Resized ROI shape:", roi_resized.shape)

```

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

hue_channel = img_hsv[:, :, 0]

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(img_rgb)

axes[0].set_title("Original Image (RGB)")

axes[0].axis("off")

axes[1].imshow(hue_channel, cmap="hsv")

axes[1].set_title("Hue Channel")

axes[1].axis("off")

plt.tight_layout()

plt.show()

```

```python
roi_resized = cv2.resize(roi, (256, 256), interpolation=cv2.INTER_LINEAR)
plt.figure(figsize=(4, 4))
plt.imshow(roi_resized)
plt.title("Resized ROI (256x256)")
plt.axis("off")
plt.show()
print("Resized ROI shape:", roi_resized.shape)
```

```python

cv2.imwrite("roi_cropped.png", cv2.cvtColor(roi, cv2.COLOR_RGB2BGR))

cv2.imwrite("roi_resized_256.png", cv2.cvtColor(roi_resized, cv2.COLOR_RGB2BGR))

cv2.imwrite("hue_channel.png", hue_channel)

print("Saved: roi_cropped.png, roi_resized_256.png, hue_channel.png")

```
