```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('Perfume.jpg', cv2.IMREAD_GRAYSCALE)
high_contrast_img = cv2.convertScaleAbs(img, alpha=1.2, beta=10)
Perfume_contrast_img = cv2.convertScaleAbs(img, alpha=0.5, beta=-30)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(high_contrast_img, cmap="gray")
plt.title("high contrast")
plt.axis("off")


plt.subplot(1, 3, 3)
plt.imshow(Perfume_contrast_img, cmap="gray")
plt.title("Perfume contrast")
plt.axis("off")

plt.show()
```

```python
# Compute histogram
hist = cv2.calcHist([img], [0], None, [256], [0, 256])

# Plot using Matplotlib
plt.figure(figsize=(10, 4))
plt.title("Image Histogram")
plt.xlabel("Intensity Value")
plt.ylabel("Pixel Count")
plt.plot(hist, color='black')
plt.xlim([0, 256])
plt.show()
```

```python
equalized_img = cv2.equalizeHist(img)
e_i = cv2.calcHist([equalized_img], [0], None, [256], [0, 256])
```

```python
plt.figure(figsize=(10, 4))
plt.title("Image Histogram")
plt.xlabel("Intensity Value")
plt.ylabel("Pixel Count")
plt.plot(e_i, color='black')
plt.xlim([0, 256])
plt.show()
```

```python
color_img = cv2.imread('color.jpg')
ycrcb = cv2.cvtColor(color_img, cv2.COLOR_BGR2YCrCb)
ycrcb[:, :, 0] = cv2.equalizeHist(ycrcb[:, :, 0])

final_color_img = cv2.cvtColor(ycrcb, cv2.COLOR_YCrCb2BGR)
```

```python
cv2.imshow('Equalized', final_color_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
